## **Uso de BAGEL4 en la terminal**

---
Hago este código para saber como emplear BAGEL4 desde la terminal.

#### ***1. Crear el ambiente con la paquetería***
``` bash
conda create -n bagel4 python=3.8 -y

conda activate bagel4
```

### ***2. Instalar las dependencias viales CONDA***
``` bash
conda config --env --add channels defaults
conda config --env --add channels conda-forge
conda config --env --add channels bioconda
conda install hmmer blast perl -y
```

### ***3. Clonar el repositorio de BAGEL4***
``` bash
git clone https://github.com/annejong/BAGEL4.git

cd BAGEL4
```

#### ***4. Instalar BioPerl y el módulo DBI***
``` bash
conda install -c bioconda perl-bioperl -y
conda install -c bioconda -c conda-forge perl-dbi perl-dbd-sqlite perl-json -y
```

#### ***5. Correr el main script para los parámetros***
``` bash
perl -Ilib bagel4_wrapper.pl
```
> *Nota: en Para Mac, se atora a partir del comando anterior, te aparece la leyenda de no encuentrar bioperl a pesar de estar instalado, por lo que tienes que hacer estos pasos:*
``` bash

find /opt/homebrew/Caskroom/miniconda/base/envs/bagel4 -name "Seq.pm"

perl -I/opt/homebrew/Caskroom/miniconda/base/envs/bagel4/lib/perl5/site_perl/5.22.0 -e 'use Bio::Seq; print "¡BioPerl funciona perfectamente!\n";'

conda install -c conda-forge -c bioconda perl-libwww-perl -y

# Esta última es para verificar los parámetros, si te los arroja es que todo se hizo correctamente:
perl -Ilib -I/opt/homebrew/Caskroom/miniconda/base/envs/bagel4/lib/perl5/site_perl/5.22.0 bagel4_wrapper.pl
```

#### ***6. Exportar PER5LIB para evitar errores en el módulo***
``` bash
export PERL5LIB="$(pwd)/lib:$PERL5LIB"
```

#### ***7. Modificar las rutas del script bagel4_wrapper.pl***
Cuando intenté correr el script con los parámetros iniciales marcaba error por la ruta donde se encontraba la carpeta BAGEL4, ya que en el script por defecto tendría que estar en `/data/bagel/lib`, por lo que debes de cambiar esa línea de comando en el script para que pueda correr sin arrojar errores. Puedes corregirlo con `nano`.
Como bagel está instalado en la oficina de la compu, la ruta corregida es:
``` bash
use lib "/mnt/c/Users/MicroIQ/Downloads/BAGEL4/lib"
```

### ***8.Correr el script (para una muestra)***
> *Nota: debido a que este programa fue instalado en la computadora de la oficina los siguientes pasos se tienen que hacer en el home directo de la terminal de ubuntu. Los fastas de los clados están copiados en el home tambien.*
``` bash
mkdir -p ~/bagel4_session/queryfolder

cp ~/fastas_clado_A/*.fna ~/bagel4_session/queryfolder/
```
