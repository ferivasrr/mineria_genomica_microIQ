## **Instalación de RODEO2 y PROKKA desde la línea de comandos**

---

Hago este instructivo para instalar RODEO2 desde la terminal para utilizarlo en la predicción de ORF para RiPP. Aunque existe el repositorio en GitHub (https://github.com/nelsonken/rodeo2) la instalación puede ser complicada debido a las versiones de python y biopython. 

La instalación de prokka es para obtener un archivo .gbk con anotaciones detalladas del genoma de los organimos, ya que RODEO2 no acepta como entrada archivos fasta, sólo .gbk

#### ***1. Clonar el repositorio de RODEO2 en tu computadora local***
``` bash
git clone https://github.com/nelsonken/rodeo2.git

cd rodeo2
```

#### ***2. Crear un ambiente en CONDA***
``` bash
conda create env -n rodeo2 python=3.7 -y

conda activate rodeo2
```

#### ***2. Instalación de las dependencias***
Instalarlas sólo en caso de no tenerlas

``` bash
conda config --env --add channels defaults
conda config --env --add channels conda-forge
conda config --env --add channels bioconda
conda install biopython=1.77 numpy scikit-learn hmmer meme -y
```

#### ***3. Comprobar que RODEO se ha instalado correctamente***
``` bash
python rodeo_main.py --help
```

#### ***4. Instalación de PROKKA***
``` bash
conda create -n prokka_env -c conda-forge -c bioconda prokka -y

conda activate prokka_env
````
