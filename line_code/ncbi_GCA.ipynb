## Descarga de secuencias GenBank (GCA_...)de NCBI desde la terminal 
 
 ---

Hago este pipeline para descargar secuencias GCA_... a partir de IDs GCF_... de la metadata de mis secuencias YcaO en el archivo `metadata_ncbi_completa_YcaO.csv`

#### ***1. Crear, activar el ambiente conda e instalar el dataset del ncbi***
``` bash
conda create -n ncbi_tools -c conda-forge ncbi-datasets-cli -y

conda activate ncbi_tools

conda install -c conda-forge ncbi-datasets-cli jq -y

datasets --version
jq --version
```

#### ***2. Cambiar los prefijos de las secuencias ***
En mi caso ya tenía un documento con los IDs (GCA_...) de mis secuencias de interés en un .txt

``` bash
sed 's/^GCF_/GCA_/' ids_assembly_clado_A.txt > gca_clado_A.txt

# Para confimrar que sí se realizó el cambio 
cat GCA_clado_A.txt
```

#### ***3. Descargar las secuencias GCA_... ***
``` bash
datasets download genome accession --inputfile GCA_clado_A.txt --include genome --filename genomas_clado-A_fasta.zip
```
#### ***4. Descomprimir el zip y poner todos los fasta en una sola carpeta ***
``` bash
unzip genomas_clado-A_fasta.zip -d datos_clado_A

mkdir -p fastas_clado_A

find datos_clado_A -type f -name "*.fna" -exec cp {} fastas_clado_A/ \;
```
